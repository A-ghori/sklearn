```python
import numpy as np
import pandas as pd
```

```python
import random
import string
import pandas as pd

df = pd.read_csv('/content/Titanic-Dataset.csv',usecols=['Survived','Ticket','Cabin'])

values = [
    random.choice(
        list(range(10)) + list(string.ascii_lowercase)
    )
    for _ in range(len(df))
]

df['new_col'] = values
df.head(10)
```

```python
df['new_col'].unique()
```

```python
## extract numerical part
df['new_col_numerical'] = pd.to_numeric(df['new_col'], errors='coerce',downcast='integer')
df['new_col_categorical'] = np.where(df['new_col_numerical'].isnull(),df['new_col'], np.nan)
df.head(2)
```

```python
df['Cabin'].unique()
```

```python
df['Cabin_num'] = df['Cabin'].str.extract('(\d+)')
df['Cabin_cat'] = df['Cabin'].str[0]
df.head(10)
```

```python
df['Ticket'].unique()
```

```python
from os import error
## Extract Ticket
df['Ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])
df['Ticket_num'] = pd.to_numeric(df['Ticket_num'],
                                 errors='coerce',
                                 downcast='integer')


df['Ticket_Cat'] = df['Ticket'].apply(lambda s: s.split()[0])
df['Ticket_Cat'] = np.where(df['Ticket_Cat'].str.isdigit(),np.nan,df['Ticket_Cat']) # condition, value_if_true, value_if_false
df
```
