## Now implement BATCH GD from scratch for multilinear regression problem

```python
class GDRegressor:

  def __init__(self,learning_rate=0.01, epochs=100):

    self.coeff = None
    self.intercept = None
    self.lr = learning_rate
    self.epochs = epochs

  def fit(self, X_train, y_train):
    # init your coeff
    self.intercept = 0
    self.coeff= np.ones(X_train.shape[1])

    print("Intercept",self.intercept)
    print("Coeff",self.coeff)
    print("Coeff Shape is", self.coeff.shape)

    for i in range(self.epochs):
      # update all the coeff and intercepts

      # intercept updation
      y_hat = np.dot(X_train, self.coeff) + self.intercept
      print("Y_hat shape is", y_hat.shape)
      intercept_der = -2 * np.mean(y_train - y_hat)
      self.intercept = self.intercept - (self.lr * intercept_der)
      print("Intercept gradient:", intercept_der)
      print("Updated intercept:", self.intercept)

      # coeff updation
      n = X_train.shape[0]
      coeff_der = -2 * np.dot((y_train - y_hat), X_train) // n
      self.coeff = self.coeff - (self.lr * coeff_der)

  def predict(self, X_test):
    return np.dot(X_test, self.coeff) + self.intercept
```

```python
gdr = GDRegressor(epochs=100)
gdr.fit(X_train,y_train)
```
