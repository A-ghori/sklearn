```python
## See that when I run date.info i can see that date is stored as an object so i cant extract month,day,year etc from this date object so i need to convert this object into actual date
date['date'] = pd.to_datetime(date['date'])
date.info()
```

```python
time['date'] = pd.to_datetime(time['date'])
time.info()
```

## How can you extract year from a date

```python
date['date_year'] = date['date'].dt.year
date.head(2)
```

## How can you extract Month from a date

```python
date['date_month'] = date['date'].dt.month
date.head(2)
```

## How can you extract Day from a date

```python
date['date_day'] = date['date'].dt.day
date.head(2)
```

## How can you extract Month_Name from a date

```python
date['date_month_name'] = date['date'].dt.month_name()
date.head(2)
```

## How can you extract week of that date

```python
date['date_week'] = date['date'].dt.day_of_week
date.head(2)
```

## day of the week - name

```python
date['date_name'] = date['date'].dt.day_name()
date.head(2)
```

## is weekend ?

```python
date['date_weekend'] = np.where(date['date_name'].isin(['Sunday','Saturday']),1,0)
date.drop(columns=['product_id','city_id','orders']).head()
```

# Extract week of the year

```python
date['date_week'] = date['date'].dt.isocalendar().week
date.head(2)
```

## extract quater

```python
date['quarter'] = date['date'].dt.quarter
date.head(2)
```
