# 2. Classification Tree — Entropy & ID3

### Entropy
- Mathematical intuition
- Formula:
  
  H(S) = -Σ pᵢ log₂(pᵢ)

- Pure vs impure nodes

### Information Gain
- Parent entropy
- Candidate split
- Weighted child entropy
- Information Gain

  IG = H(parent) - H(after split)

### Split Selection
- Evaluate possible features/splits
- Select maximum Information Gain
- Recursively repeat
